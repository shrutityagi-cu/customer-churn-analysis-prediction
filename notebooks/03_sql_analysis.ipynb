```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("../sql/churn_analysis.db")
```

```python
query = "SELECT COUNT(*) AS total_customers FROM customers;"

result = pd.read_sql_query(query, connection)
result
```

```python
query = """
SELECT
    COUNT(*) AS total_customers
FROM customers;
"""

result = pd.read_sql_query(query, connection)

result
```

```python
query = """
SELECT
    Churn,
    COUNT(*) AS customer_count,
    ROUND(
        100.0 * COUNT(*) / (SELECT COUNT(*) FROM customers),
        2
    ) AS percentage
FROM customers
GROUP BY Churn;
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT
    Contract,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        100.0 *
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY Contract
ORDER BY churn_rate DESC;
"""

contract_analysis = pd.read_sql_query(query, connection)

contract_analysis
```

```python
query = """
SELECT
    CASE
        WHEN tenure <= 12 THEN '0-12 months'
        WHEN tenure <= 24 THEN '13-24 months'
        WHEN tenure <= 48 THEN '25-48 months'
        ELSE '49+ months'
    END AS tenure_group,

    COUNT(*) AS total_customers,

    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        AS churned_customers,

    ROUND(
        100.0 *
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS churn_rate

FROM customers

GROUP BY tenure_group
ORDER BY churn_rate DESC;
"""

tenure_analysis = pd.read_sql_query(query, connection)

tenure_analysis
```

```python
query = """
SELECT
    PaymentMethod,
    COUNT(*) AS total_customers,

    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        AS churned_customers,

    ROUND(
        100.0 *
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS churn_rate

FROM customers

GROUP BY PaymentMethod

ORDER BY churn_rate DESC;
"""

payment_analysis = pd.read_sql_query(query, connection)

payment_analysis
```

```python
query = """
SELECT
    COUNT(*) AS high_risk_customers,

    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        AS churned_customers,

    ROUND(
        100.0 *
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS churn_rate

FROM customers

WHERE Contract = 'Month-to-month'
AND tenure <= 12
AND MonthlyCharges > 70;
"""

high_risk = pd.read_sql_query(query, connection)

high_risk
```
